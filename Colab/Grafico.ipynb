```python
import pandas as pd

df = pd.read_csv('Ordenação.csv', encoding='latin1')
display(df.head(20))
```

```python
df_group = (
    df.groupby(["n", "Distribuicao", "Algoritmo"], as_index=False)
      ["TempoMedio(ms)"].mean()
)


import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.barplot(
    data=df_group,
    x="Distribuicao", y="TempoMedio(ms)", hue="Algoritmo"
)
plt.title("Comparação de desempenho por tipo de distribuição")
plt.ylabel("Tempo Médio (ms)")
plt.xlabel("Distribuição dos dados")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.legend(title="Algoritmo")
plt.show()
```
